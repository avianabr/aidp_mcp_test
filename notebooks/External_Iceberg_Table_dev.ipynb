# Iceberg no bucket escolhido
Cria e registra uma tabela Iceberg usando LOCATION em um prefixo exclusivo de `oci://aidp_test@idzvuvikb5ym/aidp_codex_tests/iceberg_external/`, fora dos volumes e tabelas existentes.
Valida INSERT, MERGE, evolução, snapshots, time travel e localização efetiva. O tipo registrado deve ser confirmado pela API AIDP após executar.
Nenhum DROP é executado; dados e metadados permanecem disponíveis.


## 1. Choose a Standard catalog

Set `CATALOG_NAME` to an existing Standard catalog in your AIDP environment. The generated schema keeps this run separate from other examples.

```python
import json
import re
from uuid import uuid4

if 'spark' not in globals() or 'oidlUtils' not in globals():
    raise RuntimeError('Execute este notebook no runtime AIDP com compute anexado.')

def parameter(name, default):
    value = oidlUtils.parameters.getParameter(name, default)
    return default if value is None else str(value).strip()

def identifier(value):
    if not re.fullmatch(r'[A-Za-z_][A-Za-z0-9_]*', value):
        raise ValueError(f'Identificador inválido: {value!r}')
    return f'`{value}`'

CATALOG_NAME = parameter('CATALOG_NAME', 'default')
RUN_ID = uuid4().hex[:12]
SCHEMA_NAME = f'iceberg_ext_{RUN_ID}'
BUCKET_URI = parameter('BUCKET_URI', 'oci://aidp_test@idzvuvikb5ym')
if not re.fullmatch(r'oci://[A-Za-z0-9_.-]+@[A-Za-z0-9_-]+', BUCKET_URI):
    raise ValueError('BUCKET_URI deve ser oci://bucket@namespace.')
TABLE_LOCATION = f'{BUCKET_URI}/aidp_codex_tests/iceberg_external/{RUN_ID}/orders'
TABLE_NAME = 'orders'
TABLE = '.'.join(identifier(x) for x in (CATALOG_NAME, SCHEMA_NAME, TABLE_NAME))
result = {'run_id': RUN_ID, 'table': f'{CATALOG_NAME}.{SCHEMA_NAME}.{TABLE_NAME}', 'spark_version': spark.version}
result['requested_location'] = TABLE_LOCATION
print(json.dumps(result))
```

## 2. Create a partitioned Iceberg table

`USING iceberg` selects the Iceberg table format. The `days(order_ts)` transform partitions rows by order date without adding a separate partition column to the table.

```python
spark.sql(f"CREATE SCHEMA {identifier(CATALOG_NAME)}.{identifier(SCHEMA_NAME)}")
spark.sql(f"""
    CREATE TABLE {TABLE} (
        order_id BIGINT,
        customer_id BIGINT,
        amount DECIMAL(12, 2),
        status STRING,
        order_ts TIMESTAMP
    )
    USING iceberg
    PARTITIONED BY (days(order_ts))
    LOCATION '{TABLE_LOCATION}'
""")

spark.sql(f"DESCRIBE TABLE {TABLE}").show(truncate=False)
info = {r.col_name.strip().lower(): r.data_type for r in spark.sql(f'DESCRIBE TABLE EXTENDED {TABLE}').collect()}
actual_location = info.get('location', '').rstrip('/')
if actual_location != TABLE_LOCATION.rstrip('/'):
    raise AssertionError(f'Localização inesperada: {actual_location}')
result['actual_location'] = actual_location
result['spark_table_type'] = info.get('type', 'UNKNOWN')
```

## 3. Insert and query orders

The insert commits the first data snapshot. Save its ID so that a later step can query the table as it was before the upsert.

```python
spark.sql(f"""
    INSERT INTO {TABLE} VALUES
        (1, 10, 120.00, 'NEW', TIMESTAMP '2026-09-01 10:00:00'),
        (2, 11, 80.50, 'NEW', TIMESTAMP '2026-09-02 11:30:00')
""")

INITIAL_SNAPSHOT_ID = spark.sql(f"""
    SELECT snapshot_id
    FROM {TABLE}.snapshots
    ORDER BY committed_at ASC
    LIMIT 1
""").first().snapshot_id

spark.sql(f"SELECT order_id, customer_id, amount, status FROM {TABLE} ORDER BY order_id").show()
```

## 4. Evolve the schema and partition specification

Add a nullable `channel` column. Existing rows read as `NULL` in that column. Then add a bucket transform for `customer_id`; subsequent writes use the updated partition specification.

```python
spark.sql(f"ALTER TABLE {TABLE} ADD COLUMN channel STRING")
spark.sql(f"ALTER TABLE {TABLE} ADD PARTITION FIELD bucket(16, customer_id)")

spark.sql(f"SELECT order_id, channel FROM {TABLE} ORDER BY order_id").show()
```

## 5. Upsert with `MERGE INTO`

Match on `order_id`: update the amount and status of order `2`, and insert order `3`. The source rows also populate the new `channel` column.

```python
spark.sql(f"""
    MERGE INTO {TABLE} AS target
    USING (
        SELECT * FROM VALUES
            (2, 11, CAST(95.00 AS DECIMAL(12, 2)), 'PAID',
             TIMESTAMP '2026-09-02 11:30:00', 'web'),
            (3, 12, CAST(40.00 AS DECIMAL(12, 2)), 'NEW',
             TIMESTAMP '2026-09-03 12:00:00', 'store')
        AS source(order_id, customer_id, amount, status, order_ts, channel)
    ) AS source
    ON target.order_id = source.order_id
    WHEN MATCHED THEN UPDATE SET
        target.amount = source.amount,
        target.status = source.status,
        target.channel = source.channel
    WHEN NOT MATCHED THEN INSERT
        (order_id, customer_id, amount, status, order_ts, channel)
        VALUES (source.order_id, source.customer_id, source.amount,
                source.status, source.order_ts, source.channel)
""")

orders = spark.sql(f"""
    SELECT order_id, amount, status, channel
    FROM {TABLE}
    ORDER BY order_id
""")
orders.show()

actual = [(row.order_id, str(row.amount), row.status, row.channel)
          for row in orders.collect()]
assert actual == [
    (1, '120.00', 'NEW', None),
    (2, '95.00', 'PAID', 'web'),
    (3, '40.00', 'NEW', 'store'),
], actual
```

## 6. Inspect snapshots and query an earlier version

Each data write creates a snapshot. Compare the current table with the snapshot saved immediately after the first insert.

```python
spark.sql(f"""
    SELECT snapshot_id, committed_at, operation
    FROM {TABLE}.snapshots
    ORDER BY committed_at
""").show(truncate=False)

before_merge = spark.sql(f"""
    SELECT order_id, amount, status
    FROM {TABLE} VERSION AS OF {INITIAL_SNAPSHOT_ID}
    ORDER BY order_id
""")
before_merge.show()
assert before_merge.count() == 2
assert spark.sql(f"SELECT COUNT(*) AS row_count FROM {TABLE}").first().row_count == 3
expected_before_merge = [(1, '120.00', 'NEW'), (2, '80.50', 'NEW')]
actual_before_merge = [(r.order_id, str(r.amount), r.status) for r in before_merge.collect()]
if actual_before_merge != expected_before_merge:
    raise AssertionError(f'Time travel retornou valores inesperados: {actual_before_merge}')
result['initial_snapshot_id'] = int(INITIAL_SNAPSHOT_ID)
result['historical_rows'] = 2
result['current_rows'] = 3
result['merge_values_verified'] = True
result['time_travel_values_verified'] = True
```

## 7. Inspect the current data files

Iceberg metadata tables expose information about the files in the current snapshot. Summarize their format and row counts without opening the files directly.

```python
spark.sql(f"""
    SELECT file_format, COUNT(*) AS data_files,
           SUM(record_count) AS records,
           SUM(file_size_in_bytes) AS bytes
    FROM {TABLE}.files
    GROUP BY file_format
""").show()
```

## Retenção
O schema exclusivo e a tabela permanecem para inspeção. O armazenamento é escolhido pelo parâmetro BUCKET_URI. Não modificar os arquivos Iceberg diretamente.


```python
snapshots = spark.sql(f'SELECT snapshot_id, operation FROM {TABLE}.snapshots').collect()
if len(snapshots) < 2:
    raise AssertionError('Esperados pelo menos dois snapshots após INSERT e MERGE.')
result['snapshot_count'] = len(snapshots)
result['schema_evolution_verified'] = 'channel' in spark.table(TABLE).columns
if not result['schema_evolution_verified']:
    raise AssertionError('A coluna channel não está presente após evolução do schema.')
result['status'] = 'SUCCESS'
print('AIDP_EXTERNAL_ICEBERG_TEST_RESULT=' + json.dumps(result, sort_keys=True))
```
